```python
import pandas as pd
import yfinance as yf

# 1. Tickers: 10-Yr US Treasury Yield (^TNX), S&P 500 (^GSPC), Gold (GC=F), USD/MYR (MYR=X)
tickers = {
    '10Y_Treasury_Yield': '^TNX',
    'SP500_Index': '^GSPC',
    'Gold_Futures': 'GC=F',
    'USD_MYR': 'MYR=X'
}

# 2. Extract 2 years of historical daily market data
data_frames = []
for name, ticker in tickers.items():
    raw = yf.download(ticker, period="2y", interval="1d", progress=False)['Close']
    df_temp = pd.DataFrame(raw)
    df_temp.columns = [name]
    data_frames.append(df_temp)

# 3. Merge, clean, and forward-fill missing market dates
market_data = pd.concat(data_frames, axis=1).ffill().dropna()

# 4. Calculate Key Financial Metrics
market_data['SP500_Daily_Return_%'] = market_data['SP500_Index'].pct_change() * 100
market_data['SP500_30D_Volatility'] = market_data['SP500_Daily_Return_%'].rolling(30).std()

# 5. Export directly to formatted Excel workbook
output_filename = "JPMorgan_Macro_Market_Dashboard.xlsx"
with pd.ExcelWriter(output_filename, engine='openpyxl') as writer:
    market_data.tail(250).to_excel(writer, sheet_name="Market_Data_Summary")

print(f"Extraction complete! File saved as: {output_filename}")
```
